# **Project Title : Shopping Cart System**


```python
Scenario:
* You are creating a simple billing system where customers can enter the prices of
items they want to purchase. The program should keep accepting prices until the
user decides to stop, then apply appropriate discounts before generating the final bill.
```

Tasks:
* Continuously accept item prices using a while loop.
* Stop taking input when the user enters 0 (sentinel value).
* Keep track of the total number of items and the subtotal amount.
* Ensure no negative prices are allowed (ask again without counting as an item).

Apply discounts based on the subtotal:
* 20% for totals ≥ 1000
* 10% for totals ≥ 500
* 5% for totals ≥ 200
* 0% for totals < 200

* Display a clear bill summary including subtotal, discount amount, and final total.

```python
item_count = 0
sub_total = 0

while True:
   price = float(input("Enter item price (0 to stop): "))
   if price == 0:
        break
   elif price < 0:
        print("Negative price not allowed. Please enter again.")
        continue

   item_count += 1
   sub_total += price


# Apply discount
if sub_total >= 1000:
    discount_rate = 0.20
elif sub_total >= 500:
    discount_rate = 0.10
elif sub_total >= 200:
    discount_rate = 0.05
else:
    discount_rate = 0

discount_amount = sub_total * discount_rate
final_total = sub_total - discount_amount

# Display bill
print("\n ----- BILL SUMMARY -----")
print(f"Total items: {item_count}")
print(f"Sub_total: ₹{sub_total:.2f}")
print(f"Discount: ₹{discount_amount:.2f} ({int(discount_rate*100)}%)")
print(f"Final Total: ₹{final_total:.2f}")
```
